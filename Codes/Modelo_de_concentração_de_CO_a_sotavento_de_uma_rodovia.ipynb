## Exemplo: Concentrações de CO ao nível do solo a sotavento de uma rodovia movimentada


Em uma rodovia movimentada, um total de 10.000 veículos passam por um determinado ponto por hora em cada direção, viajando a uma velocidade média de 80 km/h. Suponha que o fator médio de emissão de CO para esses veículos seja de 20 g por milha. Suponha que prevaleçam condições moderadamente estáveis (classe F), com um vento lateral constante de 2 m/s. Determine a concentração de CO em função da distância a sotavento da rodovia. Compare os resultados com o padrão de qualidade do ar apropriado.

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Modelo de concentração de CO a sotavento de uma rodovia
#
# C = 2.04e-3 * x^(-0.6584 + 0.05367 ln(x))
#
# x : distância (km)
# C : concentração (g/m³)
# ============================================================

# Vetor de distâncias (escala logarítmica)
x = np.logspace(-1, 1, 500)   # 0.1 km até 10 km

# Equação do modelo
C_g = 2.04e-3 * x**(-0.6584 + 0.05367 * np.log(x))

# Conversão para mg/m³
C_mg = C_g * 1000

# ============================================================
# Figura
# ============================================================

plt.figure(figsize=(9, 5.5))

# Curva principal
plt.plot(
    x,
    C_mg,
    color='black',
    linewidth=2.5
)

# Linha do padrão de qualidade do ar
plt.axhline(
    y=10,
    color='black',
    linestyle='--',
    linewidth=1.5
)

# Texto do padrão
plt.text(
    0.45,
    10.7,
    'Padrão de qualidade do ar ambiente (8 h)',
    fontsize=12,
    bbox=dict(
        facecolor='white',
        edgecolor='black',
        boxstyle='round,pad=0.4'
    )
)

# Escala log no eixo x
plt.xscale('log')

# Limites dos eixos
plt.xlim(0.1, 10)
plt.ylim(0, 15)

# Ticks do eixo x
plt.xticks(
    [0.1, 0.3, 1, 3, 10],
    ['0.1', '0.3', '1', '3', '10']
)

# Labels
plt.xlabel('Distância a sotavento da rodovia (km)', fontsize=14)
plt.ylabel('Concentração de CO (mg m$^{-3}$)', fontsize=14)

# Estilo semelhante ao da figura original
plt.tick_params(
    axis='both',
    which='major',
    labelsize=12,
    direction='in',
    top=True,
    right=True
)

# Grade leve
plt.grid(False)

# Ajuste do layout
plt.tight_layout()

# Salvar figura
plt.savefig(
    'co_downwind_concentration.png',
    dpi=600,
    bbox_inches='tight'
)

plt.savefig(
    'co_downwind_concentration.pdf',
    dpi=600,
    bbox_inches='tight'
)
plt.show()
```
